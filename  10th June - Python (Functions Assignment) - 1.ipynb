In Python, functions can be categorized into built-in functions and user-defined functions. Here's a breakdown of the differences and examples for each:

Built-in Functions
Definition: Built-in functions are functions that are provided by Python itself as part of its standard library. They are always available for use in any Python program without the need for any import statements.

Examples:

print(): Outputs text to the console.

```python
print("Hello, World!") # Output: Hello, World!
```

```python
length = len("Python")
print(length)
```

```python
print(type(42)) # Output: <class 'int'>
```

User-defined Functions
Definition: User-defined functions are functions that you create yourself. They are defined using the def keyword and are specific to the program you are working on. These functions help in organizing code into reusable blocks.

Example:

A simple function to add two numbers:

```python
def add_numbers(a,b):
    return a + b

result = add_numbers (5,3)
print(result)
```

A function that prints a greeting message:

```python
def greet(name):
    print(f"Hello, {name}!")
    
    
greet("Alice")
```

Summary of Differences
Availability:

Built-in Functions: Predefined by Python and available globally.
User-defined Functions: Created by the programmer and available only after their definition.
Creation:

Built-in Functions: Defined by the Python language itself.
User-defined Functions: Defined by the programmer using the def keyword.
Usage:

Built-in Functions: Used directly without any additional setup.
User-defined Functions: Must be defined in the code before they can be used.




# 2. How can you pass arguments to a function in Python? Explain the difference between positional arguments and keyword arguments.

In Python, you can pass arguments to functions in several ways, including positional arguments, keyword arguments, and more advanced techniques like default arguments and variable-length arguments. Here's a breakdown of the different methods for passing arguments and the differences between positional and keyword arguments:

Passing Arguments to Functions
Positional Arguments

Definition: Positional arguments are the most common way to pass arguments to a function. The values provided for these arguments must match the order of the parameters defined in the function.

Example:

```python
def greet (name, age):
    print(f"Hello! {name}. You are {age} years old.")
    
# calling the function with positional arguments

greet("Alice", 30)
```

Explanation: Here, "Alice" is assigned to the parameter name, and 30 is assigned to the parameter age. The order in which you pass the arguments is crucial because it determines which parameter gets which value.

* Keyword Arguments

Definition: Keyword arguments allow you to pass arguments to a function by specifying the name of the parameter along with the value. This method makes the code more readable and allows you to pass arguments in any order.

Example:

```python
def greet (name, age):
    print(f"Hello! {name}. You are {age} years old.")
    
# calling the function with keyword arguments

greet(age=30, name="Alice")
```

Explanation: In this case, name and age are explicitly named in the function call, so their order does not matter. This can enhance code clarity, especially when functions have multiple parameters.

Differences Between Positional and Keyword Arguments
Order Dependency:

Positional Arguments: The order in which arguments are passed matters. They must match the order of parameters defined in the function.
Keyword Arguments: The order of arguments does not matter. You specify which parameter each value corresponds to by using the parameter names.
Readability:

Positional Arguments: Can become less clear if there are many parameters or if the function is called with arguments in an unexpected order.
Keyword Arguments: Improve readability by clearly indicating which value is being passed to which parameter, making the code more self-explanatory.
Flexibility:

Positional Arguments: Less flexible in terms of reordering or omitting arguments, which can lead to errors if the order is not maintained.
Keyword Arguments: More flexible, allowing you to specify only certain arguments and leave others at their default values (if default values are defined).

# 3. What is the purpose of the return statement in a function? Can a function have multiple return statements? Explain with an example.

Purpose of the return Statement in a Function
The return statement in a function serves two primary purposes:

Exit the Function: When a return statement is executed, it immediately terminates the function’s execution. No further code within the function is executed after the return statement.

Send a Value Back to the Caller: The return statement allows a function to send a value back to the code that called it. This value can be used by the calling code, stored in variables, or utilized in expressions.

Multiple return Statements in a Function
Yes, a function can have multiple return statements. Multiple return statements are often used to handle different conditions or cases within the function. When a return statement is executed, the function exits at that point, so only the code up to the return statement is executed.

Examples
Example 1: Basic Usage of return

```python
def add_positive_numbers(a, b):
    if a <= 0 or b <+0:
        return "Both number must be positive."
    return a + b

# usage

result = add_positive_numbers(5,10)
print(result)
```

```python
result = add_positive_numbers(-5,10)
print(result)
```

Explanation:

The function add_positive_numbers uses return to provide different outputs based on the conditions. If either a or b is not positive, the function returns an error message and exits. Otherwise, it returns the sum of a and b.
Example 2: Multiple Return Statements Handling Different Cases

```python
def classifiy_number(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"
    
# usage
print(classifiy_number(10))
print(classifiy_number(-5))
print(classifiy_number(0))
```

Explanation:

In this example, the function classify_number has multiple return statements to handle different cases: positive numbers, negative numbers, and zero. Each return statement provides an appropriate string based on the value of num.

# 4. What are lambda functions in Python? How are they different from regular functions? Provide an example where a lambda function can be useful.

Lambda Functions in Python

Lambda functions in Python are small, anonymous functions defined using the lambda keyword. They are often used for short, throwaway functions where defining a full function using def would be cumbersome. Lambda functions can have any number of arguments but only a single expression. The result of this expression is automatically returned.

```python
lambda arguments: expression
```

arguments: A list of parameters that the lambda function takes.
    
expression: An expression that is evaluated and returned. The expression cannot contain statements or multiple expressions.

Differences from Regular Functions

Name:

Lambda Functions: Anonymous (i.e., they are not necessarily given a name). Although you can assign them to a variable.
Regular Functions: Named using the def keyword.
    
Syntax:

Lambda Functions: Use lambda followed by parameters and an expression.
Regular Functions: Use def followed by a function name, parameters, and a block of code.
    
Complexity:

Lambda Functions: Intended for simple operations and can only contain a single expression. They are less suitable for complex logic or multiple statements.
Regular Functions: Can contain multiple statements, complex logic, and can be more descriptive.
    
Readability:

Lambda Functions: More concise but may be less readable for complex operations.
Regular Functions: More explicit and readable, especially for more complex operations.
    
Example of Lambda Functions

Example 1: Using Lambda for Sorting

Suppose you have a list of tuples, and you want to sort it based on the second element of each tuple.

```python
data = [(1, 'apple'),(3,'banana'),(2,'cherry')]

# Sorting using a lambda function
sorted_data = sorted(data, key=lambda x: x[1])

print(sorted_data)
```

 # 5. How does the concept of "scope" apply to functions in Python? Explain the difference between local scope and global scope.

In Python, scope refers to the region of a program where a particular variable or name is accessible. Understanding scope is essential for managing variables and avoiding potential conflicts or unintended behavior in your code. Python has different types of scopes that dictate the visibility and lifetime of variables.

Types of Scope in Python

Local Scope

Global Scope

* Local Scope

Local scope refers to the scope of variables that are defined within a function. Variables defined in a function are accessible only within that function. Once the function execution is complete, the local variables are destroyed.

Characteristics:

Visibility: Local variables are accessible only within the function or block in which they are defined.
Lifetime: They exist only while the function is executing.
    
Shadowing: Local variables can have the same name as global variables, but the local variables take precedence within their scope.
    
Example:

```python
def my_function():
    local_var = 10 # this is a local variable
    print("Inside function:", local_var)
    
my_function()
print("Outside function:", local_var)
```

* Explanation:

local_var is defined inside my_function and is accessible only within that function. Attempting to access local_var outside the function results in an error because it is not defined in the global scope.

Global Scope

Global scope refers to the scope of variables that are defined outside of any function. These variables are accessible from anywhere in the code after their definition.

Characteristics:

* Visibility: Global variables can be accessed from any function or block of code that comes after their definition.
Lifetime: They exist for the duration of the program.
    
* Modification: Global variables can be modified inside functions if declared as global.
    
Example:

```python
global_var = 20

def my_function():
    print("Inside function:", global_var)
    
my_function()
print("outside function:", global_var)
```

# 6. How can you use the "return" statement in a Python function to return multiple values?

In Python, you can use the return statement in a function to return multiple values by returning them as a tuple. Python’s tuple packing and unpacking make it easy to return and work with multiple values from a function.

Returning Multiple Values

When you return multiple values from a function, Python implicitly packs these values into a tuple. You can then unpack the tuple into individual variables when you call the function.

Example

Function Returning Multiple Values


```python
def get_person_info():
    name= "Alice"
    age = 30
    city = "New York"
    # Returning multiple values as a tuple
    return name, age, city

# calling the function and unpacking the return tuple
person_name, person_age, person_city = get_person_info()

print(f"Name: {person_name}")
print(f"Age: {person_age}")
print(f"City: {person_city}")
```

# 7. What is the difference between the "pass by value" and "pass by reference" concepts when it comes to function arguments in Python?

In Python, the concepts of "pass by value" and "pass by reference" are often discussed in the context of how function arguments are handled. However, Python's argument-passing mechanism is a bit different from traditional pass-by-value and pass-by-reference models.

Python’s Argument Passing Model

Python uses a model known as pass-by-object-reference or pass-by-assignment. This means that when you pass an argument to a function, you are passing a reference to the object, not the actual object itself. However, whether the function can modify the object depends on the object’s mutability (whether the object can be changed or not).

Key Concepts

Pass-by-Value:

Definition: In pass-by-value, a copy of the actual value is passed to the function. Changes made to the parameter inside the function do not affect the original argument outside the function.
Behavior in Python: Python does not use strict pass-by-value. Even though it might seem like pass-by-value for immutable types, what actually happens is that Python passes references to the values, but since the values are immutable, any "modification" results in the creation of a new object.

Pass-by-Reference:

Definition: In pass-by-reference, a reference to the actual object is passed to the function. Changes made to the parameter inside the function affect the original object.

Behavior in Python: Python does not use strict pass-by-reference. While mutable objects behave similarly to pass-by-reference, Python passes a reference to the object, and modifications to mutable objects inside the function will affect the original object.

```python
def modify_immutable(x):
    x = x + " world"  # Creates a new string object

original_string = "Hello"
modify_immutable(original_string)
print(original_string)  # Output: Hello
```

```python
def modify_mutable(lst):
    lst.append(4)  # Modifies the original list

original_list = [1, 2, 3]
modify_mutable(original_list)
print(original_list)  # Output: [1, 2, 3, 4]
```

# 8. Create a function that can intake integer or decimal value and do following operations:
    
a. Logarithmic function (log x)

b. Exponential function (exp(x))

c. Power function with base 2 (2x)

d. Square root

To create a Python function that performs multiple operations—logarithmic, exponential, power, and square root—on an integer or

decimal value, you can utilize the math module, which provides functions for these operations.

Here's a function that performs the following operations:

Logarithmic function: Computes the natural logarithm (log(x)).
    
Exponential function: Computes the exponential (exp(x)).
    
Power function with base 2: Computes  2𝑥
 .
Square root: Computes the square root of x.

```python
import math
```

```python
def perform_operations(value):
    if value <= 0:
        raise ValueError("Value must be greater than 0 for logarithmic and square root operations.")
    
    # Logarithmic function (log x)
    log_x = math.log(value)
    
    # Exponential function (exp(x))
    exp_x = math.exp(value)
    
    # Power function with base 2 (2^x)
    power_2_x = math.pow(2, value)
    
    # Square root
    sqrt_x = math.sqrt(value)
    
    return {
        "log": log_x,
        "exp": exp_x,
        "power_2": power_2_x,
        "sqrt": sqrt_x
    }

# Example usage
value = 4.0  # Can be an integer or decimal
results = perform_operations(value)

print(f"Logarithm of {value}: {results['log']}")
print(f"Exponential of {value}: {results['exp']}")
print(f"2 to the power of {value}: {results['power_2']}")
print(f"Square root of {value}: {results['sqrt']}")
```

# 9. Create a function that takes a full name as an argument and returns first name and last name.

```python
def split_name(full_name):
    # Split the full name into parts based on spaces
    parts = full_name.split()
    
    # Check if the split result has at least two parts
    if len(parts) < 2:
        raise ValueError("The full name must include at least a first name and a last name.")
    
    # Assign the first and last names
    first_name = parts[0]
    last_name = parts[-1]
    
    return first_name, last_name

# Example usage
full_name = "John Doe"
first_name, last_name = split_name(full_name)

print(f"First Name: {first_name}")
print(f"Last Name: {last_name}")
```
