 Q.1. Create two int type variables, apply addition, subtraction, division and multiplications
and store the results in variables. Then print the data in the following format by calling the
variables:
First variable is __ & second variable is __.
Addition: __ + __ = __
Subtraction: __ - __ = __
Multiplication: __ * __ = __
Division: __ / __ = __  

```python
# define two integer variables

a = 10
b =  5


# Perform operations

addition = a + b
subtraction = a - b
multiplication =  a * b
division = a / b  # note : division in Python 3 returns a float

# print the results

print(f"First variable is {a} & second variable is {b}.")
print(f"Addition: {a} + {b} = {addition}")
print(f"Subtraction: {a} - {b} = {subtraction}")
print(f"Multiplication: {a} * {b} = {multiplication}")
print(f"Division: {a} / {b} = {division:.2f}")  # Formatting division to 2 decimal places 
```

# Q.4. Explain right shift operator and left shift operator with examples.

Right Shift Operator (>>)
Behavior: Shifts the bits of a number to the right by a specified number of positions. Each right shift operation divides the number by 2 and discards the remainder. For signed integers, the sign bit is preserved, which can lead to different behavior in languages with specific integer representations. In Python, integers are unbounded and the sign is preserved automatically.

Syntax: a >> b

Here, a is the integer whose bits are to be shifted, and b is the number of positions to shift.



```python
# Binary representation of 32: 00100000
num = 32

# Shift right by 2 positions
shifted = num >> 2

# Output binary representation: 00001000
print(shifted)  # Output: 8
```

Left Shift Operator (<<)
Behavior: Shifts the bits of a number to the left by a specified number of positions. Each left shift operation multiplies the number by 2. This operation can introduce zeros on the right side of the binary representation.

Syntax: a << b

Here, a is the integer whose bits are to be shifted, and b is the number of positions to shift.

Example:

```python
# Binary representation of 3: 00000011
num = 3

# Shift left by 2 positions
shifted = num << 2

# Output binary representation: 00001100
print(shifted)  # Output: 12
```
