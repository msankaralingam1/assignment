# Q.3. What are the standards and conventions followed for the nomenclature of variables in
# python to improve code readability and maintainability?

In Python, adhering to naming standards and conventions is crucial for writing clean, readable, and maintainable code. The Python community follows several conventions outlined in PEP 8, the Python Enhancement Proposal that provides style guidelines. Here are the key standards and conventions for naming variables:

1. Use Descriptive Names
Be Clear and Specific: Choose names that convey the purpose of the variable. For instance, use total_price rather than tp to make the code more understandable.
2. Naming Conventions
Variables: Use snake_case for variable names. This means using lowercase letters and separating words with underscores (e.g., user_age, number_of_items).
Constants: Use UPPER_CASE with underscores to separate words for constants (e.g., MAX_LIMIT, PI).
Functions and Methods: Use snake_case for function and method names (e.g., calculate_total, get_user_input).
Classes: Use CamelCase for class names, where each word starts with a capital letter and no underscores are used (e.g., UserProfile, DataAnalyzer).
Modules and Packages: Use snake_case for module names and package names, and avoid underscores for packages (e.g., my_module, utilities).
3. Avoid Single Character Names
Except for counters or iterators in loops (e.g., i, j), use descriptive names that make the code self-documenting.
4. Avoid Confusing Similar Names
Do not use names that are easily confused with Python's built-in functions or keywords (e.g., avoid names like list, str, class).
5. Use Meaningful Context
Provide context in the variable names if the variable's purpose is not immediately clear from its context. For example, instead of data, use user_data or order_data.
6. Prefix for Private Variables
Use a single underscore prefix (e.g., _private_variable) for internal variables or methods that are meant for internal use only. This is a convention to indicate that these variables are intended to be private.
7. Avoid Abbreviations
Unless widely recognized (e.g., URL), avoid abbreviations that might not be immediately clear to others reading the code.

# Q.5. For what purpose def keyword is used?

In Python, the def keyword is used to define a function. Functions are blocks of reusable code that perform a specific task and can be called multiple times within a program. Functions help organize code, make it more modular, and allow for code reuse.

Syntax
The basic syntax for defining a function with the def keyword is:

````python
def function_name(parameters):
    # Function body
    statements
    return value
````

Here’s a breakdown of the syntax:

def: The keyword that indicates the start of a function definition.
function_name: The name you give to the function. It should follow the naming conventions for variables (e.g., lowercase with underscores).
parameters: Optional. These are values you pass into the function, separated by commas. They act as inputs to the function.
function body: The block of code that performs the function’s task. It is indented under the function definition.
return: Optional. This keyword is used to send a result back from the function. If no return statement is present, the function returns None by default.
Example
Here’s a simple example of how to define and use a function in Python:

````python
# Define a function to greet a user
def greet(name):
    """This function greets the person with the provided name."""
    message = f"Hello, {name}!"
    return message

# Call the function and print the result
print(greet("Alice"))
````

Key Points
Function Definition: Use def to define a function, followed by the function name and parentheses containing any parameters.
Docstring: The optional string inside triple quotes (""" """) right after the function definition is known as a docstring. It is used to describe the function’s purpose and can be accessed via help(function_name) or function_name.__doc__.
Calling Functions: Once defined, you can call the function by using its name followed by parentheses, passing any required arguments.
Functions defined using def help structure and simplify code, allowing you to break complex problems into smaller, manageable pieces.

A homogeneous list is a list where all elements are of the same type. For instance, a list of integers or a list of strings.

**Example:**


- `int_list` contains only integers.

A heterogeneous set is a set where elements can be of different types. Sets in Python are collections of unique items, and they can contain mixed types of elements.

**Example:**

```python
# A heterogeneous set with different data types
hetero_set = {1, "hello", 3.14, (2, 3)}

# Note: Sets cannot contain mutable elements like lists or dictionaries
```

In this example:
- `hetero_set` contains an integer (`1`), a string (`"hello"`), a float (`3.14`), and a tuple (`(2, 3)`).

### (iii) Homogeneous Tuple

A homogeneous tuple is a tuple where all elements are of the same type. For example, a tuple of integers or a tuple of strings.

**Example:**

```python
# A homogeneous tuple of integers
int_tuple = (10, 20, 30, 40, 50)

# A homogeneous tuple of strings
str_tuple = ("apple", "banana", "cherry")
```

In the above examples:
- `int_tuple` contains only integers.
- `str_tuple` contains only strings.

### Summary

- **Homogeneous List**: A list where all elements are of the same type. Example: `[1, 2, 3, 4, 5]`.
- **Heterogeneous Set**: A set where elements can be of different types. Example: `{1, "hello", 3.14, (2, 3)}`.
- **Homogeneous Tuple**: A tuple where all elements are of the same type. Example: `(10, 20, 30, 40, 50)`.

# Q.8. Explain the mutable and immutable data types with proper explanation & examples.

In Python, data types are categorized as mutable or immutable based on whether their values can be changed after they are created. Understanding the difference between mutable and immutable types is crucial for managing data effectively and avoiding bugs in your code.

### Mutable Data Types

**Definition:** Mutable data types are those whose values can be changed or modified after their creation. This means you can alter the contents of these objects without creating a new object.

**Examples of Mutable Data Types:**

1. **Lists**:
   - Lists are ordered collections that can contain items of different types. You can change, add, or remove elements from a list after it is created.

   **Example:**
   ```python
   my_list = [1, 2, 3]
   print(my_list)  # Output: [1, 2, 3]
   
   my_list[1] = 10  # Changing the value at index 1
   print(my_list)  # Output: [1, 10, 3]

   my_list.append(4)  # Adding a new element
   print(my_list)  # Output: [1, 10, 3, 4]
   
   my_list.remove(10)  # Removing an element
   print(my_list)  # Output: [1, 3, 4]
   ```

2. **Dictionaries**:
   - Dictionaries are collections of key-value pairs. You can modify the values associated with keys or add/remove key-value pairs.

   **Example:**
   ```python
   my_dict = {'a': 1, 'b': 2}
   print(my_dict)  # Output: {'a': 1, 'b': 2}
   
   my_dict['a'] = 10  # Changing the value for key 'a'
   print(my_dict)  # Output: {'a': 10, 'b': 2}
   
   my_dict['c'] = 3  # Adding a new key-value pair
   print(my_dict)  # Output: {'a': 10, 'b': 2, 'c': 3}
   
   del my_dict['b']  # Removing a key-value pair
   print(my_dict)  # Output: {'a': 10, 'c': 3}
   ```

3. **Sets**:
   - Sets are unordered collections of unique elements. You can add or remove elements from a set.

   **Example:**
   ```python
   my_set = {1, 2, 3}
   print(my_set)  # Output: {1, 2, 3}
   
   my_set.add(4)  # Adding an element
   print(my_set)  # Output: {1, 2, 3, 4}
   
   my_set.remove(2)  # Removing an element
   print(my_set)  # Output: {1, 3, 4}
   ```

### Immutable Data Types

**Definition:** Immutable data types are those whose values cannot be changed once they are created. Any operation that attempts to modify an immutable object will instead create a new object.

**Examples of Immutable Data Types:**

1. **Tuples**:
   - Tuples are ordered collections of elements. Once a tuple is created, you cannot change its elements or size.

   **Example:**
   ```python
   my_tuple = (1, 2, 3)
   print(my_tuple)  # Output: (1, 2, 3)
   
   # Attempting to modify a tuple element will raise an error
   # my_tuple[1] = 10  # TypeError: 'tuple' object does not support item assignment
   
   # However, you can create a new tuple by concatenating
   new_tuple = my_tuple + (4,)
   print(new_tuple)  # Output: (1, 2, 3, 4)
   ```

2. **Strings**:
   - Strings are sequences of characters. Once a string is created, its contents cannot be changed.

   **Example:**
   ```python
   my_string = "Hello"
   print(my_string)  # Output: "Hello"
   
   # Attempting to modify a string character will raise an error
   # my_string[0] = 'h'  # TypeError: 'str' object does not support item assignment
   
   # However, you can create a new string by concatenation
   new_string = my_string + " World"
   print(new_string)  # Output: "Hello World"
   ```

3. **Numbers**:
   - Numeric types (integers, floats) are immutable. Operations on these types create new values.

   **Example:**
   ```python
   my_number = 10
   print(my_number)  # Output: 10
   
   # Any arithmetic operation creates a new number
   new_number = my_number + 5
   print(new_number)  # Output: 15
   ```

### Summary

- **Mutable Data Types**: Lists, dictionaries, sets. Their values can be modified after creation.
- **Immutable Data Types**: Tuples, strings, numbers. Their values cannot be modified after creation.

Understanding the difference between mutable and immutable types is important for managing data, particularly when dealing with functions and object-oriented programming, as it affects how data is passed and manipulated.

# Q.9. Write a code to create the given structure using only for loop.
# *
# ***
# *****
# *******
# *********

````python
# number of rows
n = 5

# Loop through each row
for i in range(n):
    # Calculate the number of asterisks for the current row
    num_asterisks = 2 * i + 1
    #print the asterisks
    print('*' * num_asterisks)
````

# Q.10. Write a code to create the given structure using while loop.
# |||||||||
# |||||||
# |||||
# |||
# |

````python
#  initial number of vertical bars
num_bars = 8

# Loop until num_bars becomes non-positive

while num_bars > 0:
    #Print the vertical bars
    print('|' * num_bars)
    # Reduce the number of vertical bars by 2
    
    num_bars-= 2
````
